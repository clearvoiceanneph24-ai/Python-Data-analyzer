```python

def analyze_data(numbers):
    """Calculate basic statistics from a list of numbers."""

    if not numbers:
        return {
            "count": 0,
            "total": 0,
            "average": 0,
            "minimum": None,
            "maximum": None
        }

    return {
        "count": len(numbers),
        "total": sum(numbers),
        "average": round(sum(numbers) / len(numbers), 2),
        "minimum": min(numbers),
        "maximum": max(numbers)
    }


data = [10, 20, 30, 40, 50]

results = analyze_data(data)

print("PYTHON DATA ANALYZER")
print("--------------------")
for key, value in results.items():
    print(f"{key.capitalize()}: {value}")
```

```python

# ERROR HANDLING AND AUTOMATED TESTING

def validate_sales_data(df):
    required_columns = ["Product", "Sales", "Quantity"]

    for column in required_columns:
        if column not in df.columns:
            raise ValueError(f"Missing column: {column}")

    if df.empty:
        raise ValueError("Dataset is empty.")

    if df[required_columns].isnull().any().any():
        raise ValueError("Dataset contains missing values.")

    if (df["Sales"] < 0).any() or (df["Quantity"] < 0).any():
        raise ValueError("Negative values are not allowed.")

    return True


# AUTOMATED TESTS

assert validate_sales_data(df) == True
assert df["Total Revenue"].sum() == 100900
assert df["Product"].nunique() == 5

print("ALL TESTS PASSED!")
print("Data validation: SUCCESS")
print("Revenue calculation: SUCCESS")
print("Product count: SUCCESS")
```

```python

import matplotlib.pyplot as plt

# BAR CHART: Revenue per Product
plt.figure(figsize=(10, 5))
plt.bar(df["Product"], df["Total Revenue"])
plt.title("Revenue per Product")
plt.xlabel("Product")
plt.ylabel("Total Revenue")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()

# PIE CHART: Revenue Distribution
plt.figure(figsize=(7, 7))
plt.pie(
    df["Total Revenue"],
    labels=df["Product"],
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Sales Revenue Distribution")
plt.show()
```

```python

# SALES PERFORMANCE ANALYSIS

df["Total Revenue"] = df["Sales"] * df["Quantity"]

total_revenue = df["Total Revenue"].sum()

best_seller = df.loc[df["Quantity"].idxmax(), "Product"]

top_revenue = df.loc[df["Total Revenue"].idxmax(), "Product"]

print("SALES PERFORMANCE REPORT")
print("------------------------")
print("Total Revenue:", total_revenue)
print("Best Selling Product:", best_seller)
print("Highest Revenue Product:", top_revenue)

print("\nProduct Revenue:")
print(df[["Product", "Total Revenue"]])
```

```python

def validate_data(numbers):
    """Check whether all items are valid numbers."""

    if not numbers:
        return False, "The dataset is empty."

    if not all(isinstance(x, (int, float)) and not isinstance(x, bool) for x in numbers):
        return False, "Invalid data: numbers only."

    print("Data validation successful!")
    return True, "All entries are valid."


# Test the validation
valid_data = [10, 20, 30, 40, 50]
invalid_data = [10, 20, "hello", 40]

print(validate_data(valid_data))
print(validate_data(invalid_data))
```
